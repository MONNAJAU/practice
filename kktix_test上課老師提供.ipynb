```python
# 1. 下載並安裝 Google 官方 Linux 穩定版 Chrome (避開 Colab 的 Snap 限制)
!wget -q -O - https://dl-ssl.google.com/linux/linux_signing_key.pub | apt-key add -
!sh -c 'echo "deb [arch=amd64] http://dl.google.com/linux/chrome/deb/ stable main" >> /etc/apt/sources.list.d/google-chrome.list'
!apt-get update
!apt-get install -y google-chrome-stable

import time
from selenium import webdriver
from selenium.webdriver.common.by import By

# 2. 配置無頭模式參數
options = webdriver.ChromeOptions()
options.add_argument("--headless")  # 雲端環境必須無頭執行
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--disable-gpu")
options.add_argument("--disable-blink-features=AutomationControlled")

print("🚀 啟動 Google 官方 Chrome 瀏覽器 (無頭模式)...")
try:
    # 讓 Selenium Manager 自動配對下載對應版本的 chromedriver
    driver = webdriver.Chrome(options=options)

    url = "https://kktix.com/events/8476e7a2/registrations/new"
    driver.get(url)

    try:
        # 嘗試點擊關閉公告彈窗
        driver.find_element(By.CSS_SELECTOR, '[data-dismiss="modal"]').click()
    except Exception:
        pass

    time.sleep(6)

    elements = driver.find_elements(By.CLASS_NAME, "ticket-quantity")

    print(f"總共找到 {len(elements)} 個 ticket-quantity 元素：")

    for i, elem in enumerate(elements):
        print(f"第 {i+1} 個 text：'{elem.text}'")
        print(f"第 {i+1} 個 textContent：'{elem.get_attribute('textContent').strip()}'")
        print(f"第 {i+1} 個 HTML：{elem.get_attribute('outerHTML')}")
        print("-" * 30)

except Exception as e:
    print(f"❌ 執行發生錯誤: {e}")
finally:
    try:
        driver.quit()
    except Exception:
        pass
```
